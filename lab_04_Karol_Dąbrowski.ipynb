## 1. Struktury danych w bibliotece pandas

Podstawową strukturą danych w bibliotece pandas jest ramka danych, obiekt typu `DataFrame` oraz seria danych, `Series`, która stanowi pojedynczą kolumnę ramki danych. Pod tymi strukturami kryją się tablice numpy, które poznaliśmy już na zajęciach wcześniejszych. Każda kolumna w ramce danych jest określona typem danych, który jest w niej przechowywany, z uwzględnieniem sytuacji, w której możliwe jest również przechowywanie wartości brakujących (typ `Nan` - not a number).

### Zadanie 4

````python
column_names_iris = [
    'sepal_length_cm',
    'sepal_width_cm',
    'petal_length_cm',
    'petal_width_cm',
    'class'
]

iris_df_zad4 = pd.read_csv('iris.data', header=None, names=column_names_iris)

print("Typy danych i zużycie pamięci dla ramki 'iris_df_zad4' przed downcastingiem:")
iris_df_zad4.info(memory_usage='deep')

iris_df_zad4_downcasted = iris_df_zad4.copy()

print("Typy danych przed downcastingiem numerycznych kolumn:")
print(iris_df_zad4_downcasted.dtypes)

for col in iris_df_zad4_downcasted.select_dtypes(include=[np.number]).columns:
    iris_df_zad4_downcasted[col] = iris_df_zad4_downcasted[col].astype(np.float32)

print("Typy danych po downcastingowaniu do float32:")
print(iris_df_zad4_downcasted.dtypes)

print("Zużycie pamięci dla ramki 'iris_df_zad4_downcasted' (deep=True) po downcastingu:")
print(iris_df_zad4_downcasted.memory_usage(deep=True))
print("Całkowite zużycie pamięci dla ramki 'iris_df_zad4_downcasted' (deep=True):", iris_df_zad4_downcasted.memory_usage(deep=True).sum(), "bytes")

mem_original_iris = iris_df_zad4.memory_usage(deep=True).sum()
mem_downcasted_iris = iris_df_zad4_downcasted.memory_usage(deep=True).sum()
print(f"Oszczędność pamięci po downcastingu dla danych Iris: {((mem_original_iris - mem_downcasted_iris) / mem_original_iris) * 100:.2f}%")

display(iris_df_zad4_downcasted.head())
````

### Zadanie 7

````python
print("Zużycie pamięci dla ramki 'iris' (deep=False):")
print(iris.memory_usage())
print("Całkowite zużycie pamięci dla ramki 'iris' (deep=False):", iris.memory_usage().sum(), "bytes")

print("\nZużycie pamięci dla ramki 'iris' (deep=True):")
print(iris.memory_usage(deep=True))
print("Całkowite zużycie pamięci dla ramki 'iris' (deep=True):", iris.memory_usage(deep=True).sum(), "bytes")

iris_smaller = iris.copy()

print("\nTypy danych przed downcastingiem:")
print(iris_smaller.dtypes)

for col in iris_smaller.select_dtypes(include=[np.number]).columns:
    iris_smaller[col] = iris_smaller[col].astype(np.float32)

print("\nTypy danych po downcastingowaniu do float32:")
print(iris_smaller.dtypes)

print("\nZużycie pamięci dla ramki 'iris_smaller' (deep=True) po downcastingu:")
print(iris_smaller.memory_usage(deep=True))
print("Całkowite zużycie pamięci dla ramki 'iris_smaller' (deep=True):", iris_smaller.memory_usage(deep=True).sum(), "bytes")

mem_original = iris.memory_usage(deep=True).sum()
mem_downcasted = iris_smaller.memory_usage(deep=True).sum()
print(f"\nOszczędność pamięci po downcastingu: {((mem_original - mem_downcasted) / mem_original) * 100:.2f}%")
````
